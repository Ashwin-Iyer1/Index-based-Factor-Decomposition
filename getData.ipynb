```python
from dotenv import load_dotenv
import pandas as pd
import os
from massive import RESTClient
import json

load_dotenv()

```

```python
data_path = "XNAS-20260808-ARBSC78JBU"
```

```python
import databento as db, pandas as pd, glob

files = sorted(glob.glob(f"{data_path}/*.dbn.zst"))
df = pd.concat([db.DBNStore.from_file(f).to_df() for f in files])
```

```python
# df.to_pickle("8YearsData.pkl")
```

```python
indice_dict = {'XLE':"Energy", 
               'XLB':"Materials", 
               'XLI':"Industrials", 
               'XLY':"Consumer Discretionary", 
               'XLP':"Consumer Staples", 
               'XLV':"Healthcare", 
               'XLF':"Financials", 
               'XLK':"Technology", 
               'XLC':"Communication Services", 
               'XLU':"Utilities", 
               'XLRE':"Real Estate"
               }
```

```python
dates_list = []
for indice in indice_dict.keys():
    print(indice_dict[indice] + " earliest data: " + str(df[df['symbol'] == indice].index[0]))
    dates_list.append(df[df['symbol'] == indice].index[0])
    
earliest_date = max(dates_list)
```

```python
earliest_date
```

```python
df = df[df.index >= earliest_date]
```

```python
df = df[df['symbol'].isin(indice_dict.keys())]
```
